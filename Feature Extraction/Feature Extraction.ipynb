```python
import pandas as pd
data=pd.read_csv('../Cleaning Text/Cleaned IMDB Dataset.csv')
```

```python
data.head()
```

```python
# TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(max_features=3000)
X=tfidf.fit_transform(data['processed_review'])

# Matrix shape
print(f'TF-IDF Matrix Shape: {X.shape}')

# Display vocabulary words
print(f'\nFirst 20 TF-IDF Vocabulary Words: \n{tfidf.get_feature_names_out()[:20]}')
```

```python
# Encoding the Target Label
y=data['sentiment'].map({
    'positive':True,
    'negative':False
})
```

```python
import numpy as np
import scipy.sparse
np.save('y_mapped.npy',y)
scipy.sparse.save_npz('X_features.npz',X)
```
