```python
import numpy as np
import joblib

# Import model,test_y,y_pred
model=joblib.load('../Model/Model.pkl')
test_y=np.load('../Model/test_y.npy')
y_pred=np.load('../Model/y_pred.npy')
```

```python
# The Confusion Matrix
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

cm=confusion_matrix(test_y,y_pred)
labels=[[f'True Negative:{cm[0,0]}',f'False Positive:{cm[0,1]}'],
        [f'False Negative:{cm[1,0]}',f'True Positive:{cm[1,1]}']]
sns.heatmap(cm,annot=labels,fmt='',cmap='Blues',xticklabels=['Negative','Positive'],yticklabels=['Negative','Positive'],cbar=False)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Imdb-Sentiment Detection Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
# Print Performance Metrics
print(f'Test Accuracy: {accuracy_score(test_y,y_pred)}')
print(f"Precision: {precision_score(test_y, y_pred)}")
print(f"Recall: {recall_score(test_y, y_pred)}")
print(f"F1 Score: {f1_score(test_y, y_pred)}")
```
