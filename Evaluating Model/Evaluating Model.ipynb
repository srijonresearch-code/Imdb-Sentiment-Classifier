```python
import numpy as np
import joblib

# Import model,test_y,y_pred
xgb_model=joblib.load('../Model/xgb_model.pkl')
lcr_model=joblib.load('../Model/lcr_model.pkl')
test_y=np.load('../Model/test_y.npy')
y_pred_xgb=np.load('../Model/y_pred_xgb.npy')
y_pred_lcr=np.load('../Model/y_pred_lcr.npy')
```

```python
# The Confusion Matrix
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

cm_xgb=confusion_matrix(test_y,y_pred_xgb)
labels=[[f'True Negative:{cm_xgb[0,0]}',f'False Positive:{cm_xgb[0,1]}'],
        [f'False Negative:{cm_xgb[1,0]}',f'True Positive:{cm_xgb[1,1]}']]
sns.heatmap(cm_xgb,annot=labels,fmt='',cmap='Blues',xticklabels=['Negative','Positive'],yticklabels=['Negative','Positive'],cbar=False)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Imdb-Sentiment Detection Confusion Matrix (XGBRegressor)')
plt.show()
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
# Print Performance Metrics
print(f'Test Accuracy: {accuracy_score(test_y,y_pred_xgb)}')
print(f"Precision: {precision_score(test_y, y_pred_xgb)}")
print(f"Recall: {recall_score(test_y, y_pred_xgb)}")
print(f"F1 Score: {f1_score(test_y, y_pred_xgb)}")
```

```python
# The Confusion Matrix
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

cm_lcr=confusion_matrix(test_y,y_pred_lcr)
labels=[[f'True Negative:{cm_lcr[0,0]}',f'False Positive:{cm_lcr[0,1]}'],
        [f'False Negative:{cm_lcr[1,0]}',f'True Positive:{cm_lcr[1,1]}']]
sns.heatmap(cm_lcr,annot=labels,fmt='',cmap='Blues',xticklabels=['Negative','Positive'],yticklabels=['Negative','Positive'],cbar=False)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Imdb-Sentiment Detection Confusion Matrix (LogisticRegression)')
plt.show()
```

```python
# Print Performance Metrics
print(f'Test Accuracy: {accuracy_score(test_y,y_pred_lcr)}')
print(f"Precision: {precision_score(test_y, y_pred_lcr)}")
print(f"Recall: {recall_score(test_y, y_pred_lcr)}")
print(f"F1 Score: {f1_score(test_y, y_pred_lcr)}")
```
