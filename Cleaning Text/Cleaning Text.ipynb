```python
import pandas as pd
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
data=pd.read_csv('IMDB Dataset.csv')
```

```python
data.head(20)
```

```python
data.describe()
```

```python
data.info()
```

```python
# Checking for null values
data.isnull().sum()
```

```python
# Check Duplicates
data.duplicated().sum()
```

```python
# Drop Duplicates
data.drop_duplicates(keep='first',inplace=True)
```

```python
# Checking the balance of Reviews
data['sentiment'].value_counts()
```

```python
# Lowercasing the Reviews
data['review']=data['review'].str.lower()
```

```python
data['review'][0]
```

```python
import re

# Removing Tags
data['review']=data['review'].apply(lambda x: re.sub(r'<[^>]+>',' ',x))

# Removing Punctuation
data['review']=data['review'].apply(lambda x: re.sub(r'[^\w\s]','',x))
```

```python
data['review'][0]
```

```python
# Tokenization
from nltk.tokenize import word_tokenize

data['tokens']=data['review'].apply(word_tokenize)
```

```python
data['tokens'].head()
```

```python
# Removing Stopwords
from nltk.corpus import stopwords

stop_words=set(stopwords.words('english'))

data['tokens']=data['tokens'].apply(lambda x: [word for word in x if word not in stop_words])
```

```python
data['tokens'][0]
```

```python
# lemmatize tokens
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
data['tokens']=data['tokens'].apply(lambda x: [lemmatizer.lemmatize(word) for word in x])
```

```python
# Join tokens back into text
data['processed_review']=data['tokens'].apply(lambda x: ' '.join(x))
```

```python
data['processed_review'].head()
```

```python
# save cleaned data
data.to_csv('Cleaned IMDB Dataset.csv',index=False)
```
